Ejercicio 1

```python
from Emulador import Job

def fmap(key, value, context):
    partes = value.split("\t")
    id_visitante = partes[0]
    apuestas_local = int(partes[1])
    apuestas_visitante = int(partes[2])
    context.write(key, apuestas_local)
    context.write(id_visitante, apuestas_visitante)

def fcomb(key, values, context):
    total = 0
    for v in values:
        total += int(v)
    context.write(key, total)

def fred(key, values, context):
    total = 0
    for v in values:
        total += int(v)
    if total > context["X"]:
        context.write(key, "") # en caso de que pida el valor sumado en vez de "" hay que cambiarlo por total

job = Job("Apuestas/Premium", "Output/ejercicio1", fmap, fred)
job.addInputPath("Apuestas/Estandar", fmap)
job.setCombiner(fcomb)
job.setParams({"X": 4500})
print(job.waitForCompletion())
```

Ejercicio 3

```python
def fmapPlat3(key, value, context):
    partes = value.split("\t")
    ap_local = int(partes[1])
    ap_visitante = int(partes[2])
    if ap_local > context["X"] * ap_visitante:
        context.write(key, "platinium")                      # key = id_local

def fmapPrem3(key, value, context):
    partes = value.split("\t")
    ap_local = int(partes[1])
    ap_visitante = int(partes[2])
    if ap_local > context["X"] * ap_visitante:
        context.write(key, "premium")

def fred3(key, values, context):
    origenes = set()
    for v in values:
        origenes.add(v)
    if "platinium" in origenes and "premium" not in origenes:
        context.write(key, "")

job3 = Job("Apuestas/Platinium", "Output/ejercicio3", fmapPlat3, fred3)
job3.addInputPath("Apuestas/Premium", fmapPrem3)
job3.setParams({"X": 3})
print(job3.waitForCompletion())
```
